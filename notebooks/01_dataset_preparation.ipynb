```python
import os
from pathlib import Path

import pandas as pd

DATA_DIR = Path(os.environ.get("FD001_DIR", "/content/drive/MyDrive/predictive_maintenance"))

BASE_FEATURES = ["cycle", "setting_1", "setting_2", "setting_3"] + [f"sensor_{i}" for i in range(1, 22)]
KEEP = ["unit_id"] + BASE_FEATURES + ["degradation_score"]


def load_features(path):
    """Читає CSV з ознаками і прибирає дубль cycle.1, якщо він збігається з cycle."""
    df = pd.read_csv(path)
    if (df["cycle"] == df["cycle.1"]).all():
        df = df.drop(columns=["cycle.1"])
    return df


train = load_features(DATA_DIR / "new_train_49_features.csv")
val = load_features(DATA_DIR / "new_val_49_features.csv")

train[KEEP + ["RUL"]].to_csv(DATA_DIR / "train_main_26.csv", index=False)
val[KEEP].to_csv(DATA_DIR / "test_main_26.csv", index=False)

print("train:", train[KEEP + ["RUL"]].shape, "| validation:", val[KEEP].shape)
```
